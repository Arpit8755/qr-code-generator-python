## 1. Basic QR code from text

```python
import qrcode
data=input("Enter data to make QR:")
qr=qrcode.make(data)

qr.save("qr_basic.png")
print("QR code is ready")
print("Save as:png format")
```

## 2. URL QR with validation

```python
import qrcode
url=input("Enter URL to make QR:")
if url.startswith("http://")or url.startswith("https://"):
    qr=qrcode.make(url)
    qr.save("qr_url.png")
    print("QR code is ready")
else:
    print("INVALID")
```

## 3. Customised QR (error correction, box size, colour)

```python
import qrcode
url=input("Enter URL to make QR:")
if url.startswith("http://")or url.startswith("https://"):
    qr=qrcode.QRCode(
                    version=1,
                    error_correction=qrcode.constants.ERROR_CORRECT_H,
                    box_size=10,
                    border=4
                    )
    qr.add_data(url)
    qr.make()
    image=qr.make_image(fill_color="black",back_color="red")
    image.save("qr_custom.png")
    print("Qr ready")
else:
    print("Invalid")
```

## 4. UPI payment QR

```python
#create qr code for upi
import qrcode

upi_id=input("Enter UPI ID: ")
name=input("Enter name:")

upi_url= f"upi://pay?pa={upi_id}&pn={name}"

qr=qrcode.make(upi_url)
qr.save("qr_upi.png")


print("UPI QR IS READY")
```

## 5. UPI QR with fixed amount

```python
#create qr code for upi with specific amount
import qrcode

upi_id=input("Enter UPI ID: ")
name=input("Enter name:")
amount=int(input("Enter Amount"))

upi_url= f"upi://pay?pa={upi_id}&pn={name}&am={amount}"

qr=qrcode.make(upi_url)
qr.save("qr_upi_amount.png")



print("UPI QR IS READY")
```
